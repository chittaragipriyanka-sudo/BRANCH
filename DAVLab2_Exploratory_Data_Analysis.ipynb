```python
print("Python Marks")
print("Mean:",df["Python_Marks"].mean())
print("Median:",df["Python_Marks"].median())
print("Minimum:",df["Python_Marks"].min())
print("Maximum:",df["Python_Marks"].max())
print("Standard Deviation:",df["Python_Marks"].std())
```
