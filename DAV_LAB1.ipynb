```python
import numpy as np
import pandas as pd

print("Numpy version:",np.__version__)
print("Pandas version:",pd.__version__)
```

```python
marks=np.array([56,89,76,75,90])
print("marks:",marks)
print("Type:",type(marks))
```

```python
print("Number of dimensions:", marks.ndim)
print("Shape:",marks.shape)
print("Number of elements:", marks.size)
```

```python
marks=np.array([56,89,76,75,90])
print("Original:",marks)
print("Add 5:",marks+5)
print("Multiply by 2:",marks*2)
print("Divide by 2:",marks / 2)
print("Square:",marks**2)
```

```python
print("Sum:",marks.sum())
print("Mean:",marks.mean())
print("Maximum:",marks.max())
print("Minimum:",marks.min())
```

```python
#Numpy indexing
print("First marks:",marks[0])
print("Second marks:",marks[1])
print("Last marks:",marks[-1])
```

```python
#Njumpy slicing
print("Index 1 to 3:",marks[1:4])
print("First three:",marks[:3])
print("From index 2:",marks[2:])
print("Every second elemnt:",marks[::2])
```

```python
student_marks=np.array([
    [78,85,92],
    [67,88,76],
    [90,95,89]
])
print(student_marks)
print("Shape:",student_marks.shape)
```

```python
print("Mean of all values:",student_marks.mean())
print("Column-wise mean:",student_marks.mean(axis=0))
print("Row-wise mean:",student_marks.mean(axis=1))
```

```python
#pandas dataFrame
data={
    "Name": ["Anu","Rahul","Priya","Shreya"],
    "Branch": ["CSE","AIML","ISE","CSE"],
    "Marks": [85,91,76,88]

}
df=pd.DataFrame(data)
df
```

```python
print("Shape:",df.shape)
print("\nColumns:")
print(df.columns)
```

```python
#when want only 1 column
print("Marks column:")
print(df["Marks"])
```

```python
#only 2 columns
print("\nBranch and Marks:")
print(df["Branch"], ["Marks"])
```

```python
print("first row:")
print(df.iloc[0])
print("\nSecond row:")
print(df.iloc[1])
print("\nFirst three rows:")
print(df.iloc[0:3])
print("first 2 rows and columns")
print(df.iloc[0:2, 0:2])
```

```python
#dataframe slicing
print("first three rows:")
print(df.iloc[0:3])
print("row 1 to 3, Name and Marks:")
print(df.iloc[1:4, 0:3])
```

```python
#filtering data
high_scores = df[df["Marks"]> 90]
high_scores
```

```python
print("Students fron CSE:")
display(df[df["Branch"] == "CSE"])
print("students scoring 85 or above:")
display(df[df["Marks"]>=85])
```
